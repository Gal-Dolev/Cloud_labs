```python
# @title imports
import json
from firebase import firebase
from collections import Counter
import string
```

```python
# @title setting up firebase
link = 'https://tirgul3anan-default-rtdb.europe-west1.firebasedatabase.app/'
FBconn = firebase.FirebaseApplication(link,None)
```

```python
# @title #1
def add_single_word(userInput):

  word = userInput.lower().strip()
  word = word.translate(str.maketrans('', '', string.punctuation))

  if not word:
    return

  data = FBconn.get('/test/', None)

  if not data:
        return FBconn.post('/test/', {word: 1})

  for key, words in data.items():
      if word in words:
          words[word] += 1
          return FBconn.put('/test/' + key, word, words[word])

  return FBconn.post('/test/', {word: 1})
```

```python
# @title #2
def text_analysis(userInput):
  userInput = userInput.translate(str.maketrans('', '', string.punctuation))
  counts = Counter(userInput.lower().split())
  regDict = dict(counts)
  return FBconn.post('/test/',regDict)
```

```python
# @title #3
def update_word_count(userInput1, userInput2):
  if(not userInput2.isnumeric()):
    print("Did not input a number for count")
    return;
  count = int(userInput2)
  if(count < 1):
    print("Entered a non-positive number for count")
    return;

  word = userInput1.lower().strip()
  word = word.translate(str.maketrans('', '', string.punctuation))

  if not word:
    return

  data = FBconn.get('/test/', None)

  if not data:
        return FBconn.post('/test/', {word: count})

  for key, words in data.items():
      if word in words:
          return FBconn.put('/test/' + key, word, count)

  return FBconn.post('/test/', {word: count})
```

```python
# @title #4
def delete_word(userInput):
  word = userInput.lower().strip()
  word = word.translate(str.maketrans('', '', string.punctuation))

  if not word:
    return

  data = FBconn.get('/test/', None)

  if not data:
        return

  for key, words in data.items():
      if word in words:
          return FBconn.delete('/test/' + key, word)
```

```python
# @title #5
def view_words():
  records = FBconn.get('/test/',None)
  if records:
      print("Current Records:")

      for key, words in records.items():
        for word, count in words.items():
            print(f"{word} : {count}")
  else:
      print("\nNo records found")
```

```python
print("\nCommon Words Counter Menu:")
print("1. Add Single Word")
print("2. Add Text For Analysis")
print("3. Update Word Count")
print("4. Delete Word")
print("5. View All Words")
print("6. Exit")

while True:
    choice = input("\nEnter your choice (1-6): ")
    if(not choice.isnumeric()):
      print("Not a number")
      continue
    choice = int(choice)

    if(choice == 6):
      break
    elif(choice == 1):
      userInput = input("Enter A Single Word: ")
      if(len(userInput.split()) < 0):
        print("You Did Not Enter A Word")
      elif(len(userInput.split()) > 1):
        print("You Entered More Than One Word")
      else:
        add_single_word(userInput)
    elif(choice == 2):
      userInput = input("Enter Text For Analysis: ")
      record_id = text_analysis(userInput)
    elif(choice == 3):
      userInput1 = input("Enter Word To Update Count: ")
      userInput2 = input("Enter Desired Count: ")
      update_word_count(userInput1, userInput2)
    elif(choice == 4):
      userInput = input("Enter A Word For Deletion: ")
      delete_word(userInput)
    elif(choice == 5):
      view_words()
    else:
      print("Unidentified input")
```
