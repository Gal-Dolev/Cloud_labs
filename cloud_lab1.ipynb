```python
# @title Students Form
import IPython
from IPython.core.display import display_html
from ipywidgets import widgets
from IPython.display import display, HTML, clear_output
from google.colab.output import _publish as publish

publish.css("b {color: blue} body {background-color: #f2bcca}")
IPython.display.display_html("<b>foo bar</b>", raw=True)

students = []
with open('students.txt', 'r') as file:
    for line in file:
        students.append(line.split(","))

student_names = [student[0] for student in students]
student_names.insert(0, "")

student_dropdown = widgets.Dropdown(
    options=student_names,
    description='Student:'
)

info_output = widgets.Output()

# 1. Create the new input widgets
new_info_input = widgets.Text(description='Favorite show:')
add_button = widgets.Button(description='Add Field', button_style='success')

# 2. Define what happens when the button is clicked
def on_add_click(b):
    selected_name = student_dropdown.value
    new_value = new_info_input.value

    if selected_name and new_value:
        for data in students:
            if data[0] == selected_name:
                # Remove the newline from the previous last element if it exists
                data[-1] = data[-1].strip()
                # Append the new value with a newline
                if(len(data) > 4):
                  data[-1] = new_value + "\n"
                else:
                  data.append(new_value + "\n")
                break

        # Optional: Save the changes back to the text file
        with open('students.txt', 'w') as file:
            for data in students:
                # Re-join with commas and ensure clean newlines
                clean_data = [str(item).strip() for item in data]
                file.write(",".join(clean_data) + "\n")

        # Clear the input box and refresh the display
        new_info_input.value = ""
        show_student_info(selected_name)

add_button.on_click(on_add_click)

# 3. Separate the display logic so we can refresh it easily
def show_student_info(selected_name):
    with info_output:
        clear_output()
        if selected_name == "":
            return

        print(f"--- Information for {selected_name} ---")
        for data in students:
            if data[0] == selected_name:
                for i in range(1, len(data)):
                    print(f"Field {i}: {data[i].strip()}")
                break

        # 4. Display the input and button right below the text
        display(widgets.HBox([new_info_input, add_button]))

def on_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        show_student_info(change['new'])

student_dropdown.observe(on_change)
display(student_dropdown, info_output)
```
